## Part 1: Foundational NumPy Concepts

### 1. Array Creation & Core Attributes
Unlike native Python lists which can store heterogeneous items dynamically, a NumPy array (`ndarray`) is a homogeneous collection of elements stored contiguously in memory. Understanding an array's attributes helps you track data shapes during modeling pipelines.

```python
# Initializing a 2-dimensional array from a nested Python list
matrix = np.array([[1, 2, 3], [4, 5, 6]])

# Displaying the array and its foundational metadata
print("2D Array:\n", matrix)
print("Number of dimensions (ndim):", matrix.ndim)    # Returns number of axes
print("Shape of array (shape):", matrix.shape)        # Returns (rows, columns)
print("Total number of elements (size):", matrix.size) # Total elements across all dimensions
print("Data type of elements (dtype):", matrix.dtype)  # Unified data type of the contents
```

### Built-in Array Generators
Creating arrays manually becomes impractical for large scales. NumPy provides native helper functions to instantiate arrays efficiently with pre-allocated memory structures.

```python
# Pre-allocating a 2x3 matrix filled entirely with zeros (default float64)
zeros_arr = np.zeros((2, 3))
print("Zeros Array:\n", zeros_arr)

# Pre-allocating a 1D array of ones
ones_arr = np.ones(5)
print("\nOnes Array:\n", ones_arr)

# Generating a range of integers from [start, stop) with a defined step size
range_arr = np.arange(0, 11, 2)
print("\nSequence using arange:\n", range_arr)

# Generating an exact number of evenly spaced points over a specified closed interval [start, stop]
linspace_arr = np.linspace(0, 10, 5)
print("\nLinearly spaced values using linspace:\n", linspace_arr)
```

### 2. Fundamental Data Types (`dtypes`)
Memory optimization in numeric computing requires choosing the correct precision. Storing an array of binary flags as 64-bit integers wastes significant RAM; NumPy lets you cast types explicitly.

```python
# NumPy automatically infers the data type upon creation
default_float = np.array([1.0, 2.0, 3.0])
print("Default float type:", default_float.dtype)

# Explicitly specifying a downsized 32-bit floating point for memory efficiency
float_32_arr = np.array([1.5, 2.7, 3.9], dtype=np.float32)
print("Explicit 32-bit float array:", float_32_arr)

# Safely casting an existing array to integers (this truncates decimal points)
int_32_arr = float_32_arr.astype(np.int32)
print("Array casted to 32-bit integers:", int_32_arr)
```

### 3. Basic Vectorized Arithmetic
In vanilla Python, performing element-wise arithmetic requires list comprehensions or explicit `for` loops. NumPy implements **Vectorization**, which delegates operations to optimized C loops under the hood.

```python
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

# Element-wise addition without writing loops
addition_result = a + b
print("Element-wise Addition (a + b):", addition_result)

# Scalar operations apply the arithmetic change evenly across every single item
scalar_mult = a * 2
print("Scalar Multiplication (a * 2):", scalar_mult)

# Raising every item to an exponent power
exponent_result = a ** 2
print("Element-wise Squaring (a ** 2):", exponent_result)
```

## Part 2: Intermediate Operations & Array Manipulation

### 1. Indexing, Slicing, and Boolean Masking
Extracting data from multi-dimensional structures follows standard syntax patterns. **Note:** Basic slicing creates a **view**, not a copy. Changing a slice alters the original data structure unless explicitly cloned via `.copy()`.

```python
matrix_2d = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("Base 2D Matrix:\n", matrix_2d)

# Accessing a single coordinate: matrix_2d[row_index, column_index]
print("\nElement at row 0, col 2:", matrix_2d[0, 2])

# Slicing dimensions: matrix_2d[row_slice, column_slice]
# Grabs rows 0 and 1, and columns 1 and 2
sub_matrix = matrix_2d[0:2, 1:3]
print("First two rows, last two columns:\n", sub_matrix)
```

### Boolean Masking (Filtering)
Boolean masking acts as an elite search filter, matching array positions against conditional constraints to instantly extract elements.

```python
data_arr = np.array([5, 12, 18, 3, 25, 8])
print("Data Array:", data_arr)

# Evaluating a logical statement outputs a same-shaped array of True/False flags
mask = data_arr > 10
print("Boolean Mask Matrix:", mask)

# Passing the mask into the index brackets isolates matching elements
filtered_data = data_arr[mask]
print("Filtered elements (values > 10):", filtered_data)
```

### 2. Reshaping and Transposing
Reshaping alters the dimensional layout of structural data without editing underlying cell contents. The product of the target shape dimensions must always equal the original array size.

```python
flat_vector = np.arange(1, 7)
print("Original 1D Vector:", flat_vector)

# Converting a 1D vector of 6 items into a 2D matrix of shape (2, 3)
reshaped_matrix = flat_vector.reshape(2, 3)
print("Reshaped to 2x3 Matrix:\n", reshaped_matrix)

# Using -1 allows NumPy to dynamically infer the remaining dimension shape automatically
flattened = reshaped_matrix.reshape(-1)
print("Flattened back to 1D:", flattened)

# Transposing matrix orientation (axes swapped around the main diagonal index)
transposed_matrix = reshaped_matrix.T
print("Transposed Matrix (Rows swapped with Columns):\n", transposed_matrix)
```

### 3. Understanding Broadcasting Rules
Broadcasting specifies the operational rules NumPy follows to run calculations on arrays with mismatched dimensional layouts. When performing an element-wise operation on two arrays, NumPy compares their shapes element-wise from right to left. Two dimensions are compatible if they are equal, or if one of them is 1.

```python
base_matrix = np.array([[1, 2], [3, 4], [5, 6]]) # Shape: (3, 2)
row_modifier = np.array([10, 20])               # Shape: (2,)

print("Base Matrix (3, 2):\n", base_matrix)
print("Row Modifier Vector (1, 2):", row_modifier)

# The single dimension vector is stretched vertically down to match the matrix shape
broadcasted_result = base_matrix + row_modifier
print("Broadcasted Result Matrix:\n", broadcasted_result)
```

## Part 3: Practical Applications & Common Use Cases

### 1. Data Normalization (Min-Max Scaling)
Feature scaling normalizes continuous variables into bounded ranges like [0, 1]. This optimization step helps stabilize model calculations during machine learning training pipelines.

```python
raw_features = np.array([100.0, 150.0, 300.0, 120.0, 500.0])
print("Raw Feature Vector:", raw_features)

# Calculating baseline stats
min_val = raw_features.min()
max_val = raw_features.max()

# Executing vector-accelerated Min-Max equation
normalized_features = (raw_features - min_val) / (max_val - min_val)
print("Normalized Bound Vector:", normalized_features)
```

### 2. Vectorized Conditional Logic (`np.where`)
Using `np.where` acts like an optimized if-else branching loop over an entire array, substituting data points based on whether they meet specific condition constraints.

```python
signal_array = np.array([12, -5, 18, -2, 25, 0])
print("Original Signal Array:", signal_array)

# Syntax: np.where(condition, value_if_true, value_if_false)
processed_signal = np.where(signal_array > 0, "Positive", "Negative")
print("Processed Categorized Signal:", processed_signal)
```

### 3. Aggregation and Descriptive Statistics
NumPy contains mathematical aggregation methods. You can specify the `axis` parameter to calculate descriptive statistics across an entire array, or along specific rows or columns.

```python
data_matrix = np.array([[1, 2], [3, 4], [5, 6]])
print("Data Matrix Layout:\n", data_matrix)

# 1. Flattened calculation over the entire array structure
print("\nGlobal Sum (all items flattened):", data_matrix.sum())

# 2. Column-wise aggregation (collapsing rows vertically downward)
print("Column Sums (axis=0 calculation):", data_matrix.sum(axis=0))

# 3. Row-wise aggregation (collapsing columns horizontally rightward)
print("Row Means (axis=1 calculation):", data_matrix.mean(axis=1))
```

## Summary & Key Takeaways

- **Fixed Shapes & Homogeneous Types:** NumPy arrays use a fixed size and consistent data types (`dtype`) to keep memory contiguous and efficient.
- **Vectorization Eliminates Explicit Loops:** Applying math operations to an array updates every element using fast, underlying C code instead of slower Python `for` loops.
- **Slicing Generates Views:** Modifying a sliced section of an array changes the original data. If you want to keep the original intact, create an independent copy using `.copy()`.
- **Axes Drive Aggregations:** When running descriptive summary metrics, remember that `axis=0` calculates values down columns, while `axis=1` runs calculations across rows.
